```python
#Loop over a collection
colors = ['red','green','blue','yellow']
for color in colors:
    print(color)
```

```python
# Loop backwards: "reversed"
for color in reversed(colors):
    print(color)
```

```python
# Looping over a collection and indicies in the same time:
for i, color in enumerate(colors):
    print (i,'-->', color)
```

```python
# Loop over two collections: zip()
name = ['raymond','rachel','matthew']
for name, color in zip(name,colors):
        print(name,'-->',color)
```

```python
# Looping over sorted order:
for color in sorted(colors):
    print(color)
```

```python
# Looping over sorted order backwards:
for color in sorted(colors, reverse=True):
    print(color)
```

```python
# Custom Sort order：自定义排序: 按字母长度排：
# 99% of the time ： key=XXX
# Replace Comparsion function with  key function :
''' sorted colors key equal length the key function gets called
exactly once per key '''
```

```python
print(sorted(colors, key=len)) # Perfer to use key function instead of comparasion function
```

```python
# Call a function until a sentinel value (哨兵值):
''' iter(callable, sentinal): callable function must
be no paramerer function , if not, use partial to fix it '''
# get rid of while True loop, using iter(callable, sentinel) instead:
from functools import partial
import io

# Create a file-like object from a string for demonstration
f = io.StringIO("This is a test string to demonstrate reading in blocks.")

blocks = []
for block in iter(partial(f.read, 5), ''): # Read in blocks of 5 characters
    blocks.append(block)

print(blocks)
```

1. 想用 for…else，if 里必须有 break 或 return 才能体现“多种退出路径”的区别。

2. 想最简单：找到就 return，循环完再统一 return -1。

```python
# Distinguishing multiple exit points in loops:
# for ...break ... else...

def find(seq, target):
  for i, value in enumerate(seq):
    if value == target:
       break
  else:
      return -1
  return i
```

```python
# Count with dictionaries:
colors =['red','green','red','blue','green','red']
colors_2 = ['red','green','red','blue','green','red']
d={}

for color in colors:
  if color not in d:
    d[color] = 1
  else:
    d[color] += 1

print(d)
```

```python
# Count with dictionaries with 'get' :

d={}
for color in colors_2:
  d[color] = d.get(color, 0) + 1
print(d)
```

```python
# mordern way to count dictionaries:
# defaultdict(int)：缺失键默认是 0:
# 想更省事，不想手动“初始化 0”:
colors_3 = ['red','green','red','blue','green','red']
from collections import defaultdict

d=defaultdict(int)

for color in colors_3:
  d[color] +=1
print(d)
```

```python
# Group with dictionaries:
'''用字典做“分组”（grouping）——把一堆元素按某个“分组键”装进不同的桶里。
这里的“分组键”是名字的长度'''
names = ['apple','we','lu','wulala','pear','abcd']

d = {}
for name in names:
  key = len(name)
  if key not in d:
     d[key] = []
  d[key].append(name)

print(d)
```

```python
# Better ! - Use setdefault - grouping with dictionaries:
d = {}
for name in names:
  key = len(name)
  d.setdefault(key, []).append(name)
print(d)
```

```python
# or use defualtlist(int):

d = defaultdict(list)
for name in names:
  key = len(name)
  d[key].append(name)
print(d)
```
